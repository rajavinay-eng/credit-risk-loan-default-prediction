```python
import pandas as pd
import json, joblib
from pathlib import Path

b = joblib.load("models/credit_model.joblib")
features, cat_cols = b["features"], b["cat_cols"]
train = pd.read_parquet("data/train_clean.parquet")

defaults = {}
for f in features:
    if f in cat_cols:
        m = train[f].mode(dropna=True)
        defaults[f] = None if m.empty else str(m.iloc[0])
    else:
        v = train[f].median()
        defaults[f] = None if pd.isna(v) else float(v)

Path("models/feature_defaults.json").write_text(json.dumps(defaults, indent=1))
print("Saved", len(defaults), "typical values")
print({k: defaults[k] for k in ["EXT_SOURCE_2", "AMT_INCOME_TOTAL", "AMT_CREDIT", "NAME_INCOME_TYPE"]})
```
